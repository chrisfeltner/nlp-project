```python
from sentence_transformers import SentenceTransformer
import pandas as pd
```

Download embedding model (Qwen has support for Hungarian language and performed well compared to fine-tuned models on benchmarks). Using a smaller model so that we can inference on a personal computer.

```python
model = SentenceTransformer("Qwen/Qwen3-Embedding-0.6B")
```

```python
sampled_dataframe = pd.read_pickle('sampled_dataframe.pkl')
```

```python
sampled_dataframe['embeddings'] = sampled_dataframe.sentence.map(lambda x: model.encode(x))
```
